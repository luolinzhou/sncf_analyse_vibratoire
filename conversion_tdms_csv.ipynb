```python
from nptdms import TdmsFile

# Lire le fichier TDMS
tdms_file = TdmsFile.read("/Volumes/Seagate Drive/pdi/nouvelle_tournee/24091902si11/File_by_file/results_001.tdms")

# Convertir en DataFrame pandas
df = tdms_file.as_dataframe()

# Exporter en CSV
df.to_csv("/Volumes/Seagate Drive/pdi/nouvelle_tournee/24091902si11/File_by_file/results_001.csv", index=False)
```

```python
# Dossier contenant les fichiers TDMS
input_folder = "/Volumes/Seagate Drive/pdi/nouvelle_tournee/24091902si11/File_by_file/"
output_folder = input_folder  # On met les CSV dans le même dossier

# Boucle sur les fichiers de 000 à 039
for i in range(40):
    file_number = f"{i:03d}" 
    tdms_path = os.path.join(input_folder, f"results_{file_number}.tdms")
    csv_path = os.path.join(output_folder, f"results_{file_number}.csv")

    # Vérifier si le fichier existe avant de traiter
    if os.path.exists(tdms_path):
        print(f"Conversion de {tdms_path} vers {csv_path}...")
        tdms_file = TdmsFile.read(tdms_path)  
        df = tdms_file.as_dataframe() 
        df.to_csv(csv_path, index=False) 
    else:
        print(f"Fichier {tdms_path} non trouvé, passage au suivant.")

print("✅ Conversion terminée.")
```
